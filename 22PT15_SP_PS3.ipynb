#Q1

We have a Media Player application that can play only MP3 files. Now we want to support MP4 and VLC formats, but their players have different interfaces.

```python
# -- ADAPTER PATTERN --

from abc import ABC, abstractmethod

class MediaPlayer(ABC):
    @abstractmethod
    def play(self, file_name):
        pass


# existing class
class MP3Player():
    def play(self, file_name):
      print(f"\nPlaying MP3 file: {file_name}")

# New class
class MP4Player():
    def play_mp4_file(self, file_name):
      print(f"\nPlaying MP4 file: {file_name}")

class VLCPlayer():
    def play_vlc_file(self, file_name):
      print(f"\nPlaying VLC file: {file_name}")


# adapter class
class MP4PlayerAdapter(MediaPlayer):
    def __init__(self, mp4_player):
      self.mp4_player = mp4_player
    def play(self, file_name):
      self.mp4_player.play_mp4_file(file_name)

class VLCPlayerAdapter(MediaPlayer):
    def __init__(self, vlc_player):
      self.vlc_player = vlc_player
    def play(self, file_name):
      self.vlc_player.play_vlc_file(file_name)


def play_media(media_player, file_name):
    media_player.play(file_name)


mp3 = MP3Player()
mp4_adapter = MP4PlayerAdapter(MP4Player())
vlc_adapter = VLCPlayerAdapter(VLCPlayer())

play_media(mp3, "file_name.mp3")
play_media(mp4_adapter, "file_name.mp4")
play_media(vlc_adapter, "file_name.vlc")
```

#Q2

As part of home automation we need to control different appliances
such as :

a. Lights

b. Fans

c. TV, etc

These appliances need to be controlled with different types of
operations such as :

a. On or Off (toggle)

b. Increase or decrease (regulator)

```python
# -- BRIDGE PATTERN --

from abc import ABC, abstractmethod

class Appliance(ABC):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def turn_on(self):
        pass
    @abstractmethod
    def turn_off(self):
        pass
    @abstractmethod
    def increase(self):
        pass
    @abstractmethod
    def decrease(self):
        pass


# Concrete class
class Light(Appliance):
    def __init__(self):
        super().__init__("Light")
    def turn_on(self):
        print(f"{self.name} is ON")
    def turn_off(self):
        print(f"{self.name} is OFF")
    def increase(self):
        print(f"Increasing brightness of {self.name}")
    def decrease(self):
        print(f"Decreasing brightness of {self.name}")

class Fan(Appliance):
    def __init__(self):
        super().__init__("Fan")
    def turn_on(self):
        print(f"{self.name} is ON")
    def turn_off(self):
        print(f"{self.name} is OFF")
    def increase(self):
        print(f"Increasing speed of {self.name}")
    def decrease(self):
        print(f"Decreasing speed of {self.name}")

class TV(Appliance):
    def __init__(self):
        super().__init__("TV")
    def turn_on(self):
        print(f"{self.name} is ON")
    def turn_off(self):
        print(f"{self.name} is OFF")
    def increase(self):
        print(f"Increasing volume of {self.name}")
    def decrease(self):
        print(f"Decreasing volume of {self.name}")


# Abstraction
class ApplianceController:
    def __init__(self, appliance):
        self.appliance = appliance
    def turn_on(self):
        self.appliance.turn_on()
    def turn_off(self):
        self.appliance.turn_off()

# Refined abstraction
class AdvancedApplianceController(ApplianceController):
    def increase(self):
        self.appliance.increase()
    def decrease(self):
        self.appliance.decrease()


print("-- Controlling TV --")
tv = TV()
appliance_controller = ApplianceController(tv)
appliance_controller.turn_on()
appliance_controller.turn_off()

print("\n-- Controlling Light --")
light = Light()
advanced_appliance_controller = AdvancedApplianceController(light)
advanced_appliance_controller.turn_on()
advanced_appliance_controller.increase()
advanced_appliance_controller.decrease()
advanced_appliance_controller.turn_off()
```

#Q3

Implement file explorer functionality to support navigation – folders can contain files or sub-folders and so on.

```python
# -- COMPOSITE PATTERN --

from abc import ABC, abstractmethod

# Common interface
class FileSystemElement(ABC):
  @abstractmethod
  def show_details(self):
      pass
  @abstractmethod
  def get_size(self):
      pass


# Leaf class
class File(FileSystemElement):
  def __init__(self, name, size):
    self.name = name
    self.size = size

  def show_details(self, indent: int = 0):
    prefix = "  " * indent
    print(f"{prefix}-- {self.name} ({self.size} KB)")

  def get_size(self):
    return self.size


# Composite class
class Folder(FileSystemElement):
  def __init__(self, name):
    self.name = name
    self.children = []

  def add(self, element: FileSystemElement):
    self.children.append(element)

  def remove(self, element: FileSystemElement):
    self.children.remove(element)

  def get_size(self):
    return sum(child.get_size() for child in self.children)

  def show_details(self, indent: int = 0):
    prefix = "  " * indent
    print(f"{prefix}** {self.name} (Total: {self.get_size()} KB)")
    # Recurse into children
    for child in self.children:
      child.show_details(indent + 1)


# Create files
file1 = File("budget.xlsx", 200)
file2 = File("cat_photo.jpg", 1500)
file3 = File("todo.txt", 5)
file4 = File("vacation_video.mp4", 50000)

# Create folders
root = Folder("Home")
docs = Folder("Documents")
media = Folder("Media")
photos = Folder("Photos")

# Build the hierarchy
root.add(docs)
root.add(media)

docs.add(file1)
docs.add(file3)

media.add(photos)
media.add(file4)

photos.add(file2)

# Display the structure
print("--- File Explorer Navigation ---\n")
root.show_details()
```

#Q4

A video streaming platform to play videos with its core content and allow features like subtitles, different language audio tracks, video quality options, and audio enhancements etc.

```python
# --- DECORATOR PATTERN ---

from abc import ABC, abstractmethod

class VideoStreamer(ABC):
    @abstractmethod
    def stream(self):
        pass

class SimpleVideoStreamer(VideoStreamer):
    def __init__(self, video_name):
        self.video_name = video_name
    def stream(self):
        return f"Streaming: {self.video_name}"


# Base Decorator
class VideoStreamerDecorator(VideoStreamer):
    def __init__(self, video_streamer):
        self.video_streamer = video_streamer
    def stream(self):
        return self.video_streamer.stream()


# Concrete Decorators
class SubtitleDecorator(VideoStreamerDecorator):
    def stream(self):
        return f"{super().stream()}\n [+] Subtitles enabled (English)"

class AudioLanguageDecorator(VideoStreamerDecorator):
    def __init__(self, video_streamer, language):
        super().__init__(video_streamer)
        self.language = language
    def stream(self):
        return f"{super().stream()}\n [+] Audio track: {self.language}"

class VideoQualityDecorator(VideoStreamerDecorator):
    def __init__(self, video_streamer, quality):
        super().__init__(video_streamer)
        self.quality = quality
    def stream(self):
        return f"{super().stream()}\n [+] Resolution: {self.quality}"

class AudioEnhancementDecorator(VideoStreamerDecorator):
    def stream(self):
        return f"{super().stream()}\n [+] Dolby Atmos Audio enabled"


video = SimpleVideoStreamer("Video.mp4")
video = VideoQualityDecorator(video, "4K Ultra HD")
video = SubtitleDecorator(video)
video = AudioLanguageDecorator(video, "Spanish")
video = AudioEnhancementDecorator(video)
print(video.stream())
```

#Q5

A travel booking app might offer a single method for booking a complete travel package, handling the interactions with each service and presenting a simplified interface to the user.

```python
# --- FACADE PATTERN ---

# Complex Subsystems
class FlightService:
  def book_flight(self, destination):
    print(f"- Flight to {destination} booked.")

class HotelService:
  def book_hotel(self, destination):
    print(f"- Hotel room in {destination} reserved.")

class PaymentService:
  def process_payment(self, amount):
    print(f"- Payment of ${amount} processed.")


# Facade class
class TravelBookingFacade:
  def __init__(self):
    self.flight_service = FlightService()
    self.hotel_service = HotelService()
    self.payment_service = PaymentService()

  def book_travel(self, destination, amount):
    print("Starting travel booking process...")
    self.flight_service.book_flight(destination)
    self.hotel_service.book_hotel(destination)
    self.payment_service.process_payment(amount)
    print("Completed travel booking process!")

travel_booking = TravelBookingFacade()
travel_booking.book_travel("New York", 500)
```

#Q6

You’re building a text editor where:
* Each character on screen is represented by a Character object.
* Characters share common formatting like font, size, color (intrinsic
state).
* Their position (x, y) is unique (extrinsic state).

```python
# --- FLYWEIGHT PATTERN ---

# Flyweight class
class Character:
  def __init__(self, char, font, size, color):
    self.char = char
    self.font = font
    self.size = size
    self.color = color

  def display(self, x, y):
    print(f"Drawing '{self.char}' (Font: {self.font}, {self.size}pt, {self.color}) at ({x}, {y})")


# Flyweight Factory
class CharacterFactory:
  _characters = {}

  @classmethod
  def get_character(self, char, font, size, color):
    key = (char, font, size, color)
    if key not in self._characters:
      self._characters[key] = Character(char, font, size, color)
    return self._characters[key]


# Context class
class TextEditor:
  def __init__(self):
    self.characters_to_draw = []

  def add_character(self, char, font, size, color, x, y):
    character = CharacterFactory.get_character(char, font, size, color)
    self.characters_to_draw.append((character,x, y))

  def draw(self):
    for character, x, y in self.characters_to_draw:
      character.display(x, y)


editor = TextEditor()
editor.add_character("H", "Arial", 8, "Red", 10, 11)
editor.add_character("e", "Arial", 8, "Red", 10, 12)
editor.add_character("l", "Arial", 8, "Red", 10, 13) # 'l' #1 created
editor.add_character("l", "Arial", 8, "Red", 10, 14) # 'l' #2 REUSES the object above
editor.add_character("o", "Arial", 8, "Red", 10, 15)
editor.draw()
```

#Q7

We have a real application service that performs business operations. We want to add logging, monitoring, and access tracking without changing the original service code.

```python
# --- PROXY PATTERN ---

import time
from abc import ABC, abstractmethod

# Subject Interface
class BusinessService(ABC):
    @abstractmethod
    def process_data(self, user_id, data):
        pass


# Legacy code
class RealAppService(BusinessService):
    def process_data(self, user_id, data):
        print(f"Service: Processing business logic for {data}...")
        time.sleep(0.2)


# The Proxy (Adding Logging, Monitoring, and Access Tracking)
class ServiceProxy(BusinessService):
    def __init__(self, real_service):
        self._real_service = real_service

    def process_data(self, user_id, data):
        # Access tracking
        print(f"[SECURITY] User '{user_id}' requested access.")

        # Logging
        print(f"[LOGGING] Started processing '{data}' at {time.strftime('%H:%M:%S')}")

        # Monitoring (Start Timer)
        start_time = time.perf_counter()

        # Delegate the actual work to the real service
        self._real_service.process_data(user_id, data)

        # Monitoring (End Timer)
        end_time = time.perf_counter()
        print(f"[MONITOR] Process took {end_time - start_time:.4f} seconds.")
        print("-" * 40)


real_service = RealAppService()
proxy = ServiceProxy(real_service)
proxy.process_data("user_01", "Inventory_Update")
proxy.process_data("user_02", "Payment_Batch_44")
```
