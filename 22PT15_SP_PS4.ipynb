#Q1

A company develops a Smart Home Automation System where a mobile app
or voice assistant can control multiple devices:
- Lights (ON/OFF, Brightness control)
- Fan (ON/OFF, Speed control)
- Air Conditioner (ON/OFF, Temperature control)
- Door Lock (Lock/Unlock)

Users can also:
- Schedule actions
- Undo last action
- Execute macro commands (e.g., “Good Night” turns off everything)

```python
# COMMAND PATTERN

from abc import ABC, abstractmethod

class Command(ABC):
  @abstractmethod
  def execute(self):
    pass
  @abstractmethod
  def undo(self):
    pass

# Reciever class (Business logic)
class Light:
    def light_on(self):
        print("Light ON")
    def light_off(self):
        print("Light OFF")

class Fan:
    def fan_on(self):
        print("Fan ON")
    def fan_off(self):
        print("Fan OFF")

class AirConditioner:
    def ac_on(self):
        print("AC ON")
    def ac_off(self):
        print("AC OFF")

class DoorLock:
    def door_lock(self):
        print("Door Locked")
    def door_unlock(self):
        print("Door Unlocked")


# Concrete command
class LightOnCommand(Command):
    def __init__(self, light):
        self.light = light
    def execute(self):
        self.light.light_on()
    def undo(self):
        self.light.light_off()
class LightOffCommand(Command):
    def __init__(self, light):
        self.light = light
    def execute(self):
        self.light.light_off()
    def undo(self):
        self.light.light_on()


class FanOnCommand(Command):
    def __init__(self, fan):
        self.fan = fan
    def execute(self):
        self.fan.fan_on()
    def undo(self):
        self.fan.fan_off()
class FanOffCommand(Command):
    def __init__(self, fan):
        self.fan = fan
    def execute(self):
        self.fan.fan_off()
    def undo(self):
        self.fan.fan_on()

class ACOnCommand(Command):
    def __init__(self, ac):
        self.ac = ac
    def execute(self):
        self.ac.ac_on()
    def undo(self):
        self.ac.ac_off()
class ACOffCommand(Command):
    def __init__(self, ac):
        self.ac = ac
    def execute(self):
        self.ac.ac_off()
    def undo(self):
        self.ac.ac_on()

class DoorLockCommand(Command):
    def __init__(self, door):
        self.door = door
    def execute(self):
        self.door.door_lock()
    def undo(self):
        self.door.door_unlock()
class DoorUnlockCommand(Command):
    def __init__(self, door):
        self.door = door
    def execute(self):
        self.door.door_unlock()
    def undo(self):
        self.door.door_lock()


class MacroCommand(Command):
    def __init__(self, commands):
        self.commands = commands
    def execute(self):
        for command in self.commands:
            command.execute()
    def undo(self):
        for command in reversed(self.commands):
            command.undo()


# Invoker
class RemoteControl:
    def __init__(self):
        self.history = []

    def execute_command(self, command):
        command.execute()
        self.history.append(command)

    def undo(self):
        if self.history:
            command = self.history.pop()
            command.undo()


# Client code
light = Light()
fan = Fan()
ac = AirConditioner()
door = DoorLock()

light_cmd = LightOnCommand(light)
fan_cmd = FanOnCommand(fan)
ac_cmd = ACOnCommand(ac)
door_cmd = DoorUnlockCommand(door)

remote = RemoteControl()

print("Executing sequence of commands:")
remote.execute_command(light_cmd)
remote.execute_command(fan_cmd)
remote.execute_command(ac_cmd)
remote.execute_command(door_cmd)

print("\nUndo last action:")
remote.undo()

print("\nMacro command - Good night:")
good_night = MacroCommand(
    [LightOffCommand(light), FanOffCommand(fan), ACOffCommand(ac), DoorLockCommand(door)]
)
remote.execute_command(good_night)
```

#Q2

Develop a customer support chatbot for banking services.
Users type messages like:
- “Check my balance”
- “Transfer 5000 to Ravi”
- “Show last 5 transactions”
- “Block my card if lost”

The chatbot must interpret the meaning of the sentence and execute the
correct banking service

```python
# INTERPRETER PATTERN

from abc import ABC, abstractmethod

class Expression(ABC):
    @abstractmethod
    def interpret(self):
        pass

# Terminal experssions
class Balance(Expression):
    def interpret(self):
        return "Your balance is $50,000"

class Transactions(Expression):
    def __init__(self, count):
        self.count = count
    def interpret(self):
        return f"Showing last {self.count} transactions"

class BlockCard(Expression):
    def interpret(self):
         return "Your card has been blocked"


# Non-terminal Expression
class TransferFunds(Expression):
    def __init__(self, amount, recipient):
        self.amount = amount
        self.recipient = recipient
    def interpret(self):
        return f"Transferring ₹{self.amount} to {self.recipient}"


# Client Code
commands = [
    Balance(),
    TransferFunds(5000, "Ravi"),
    Transactions(5),
    BlockCard()
]
for command in commands:
    print(command.interpret())
```

#Q3

A university develops a Library Management System that stores thousands of
books across categories:
- Computer Science
- Electronics
- Mechanical
- Literature
- Journals &amp; Magazines

Different users need to traverse books in different ways:
- Student → Browse all books
- Librarian → Check issued books
- Admin → Generate inventory report
- Visitor → Browse available books only

The system must allow traversal without exposing internal storage structure.

```python
# ITERATOR PATTERN

from abc import ABC, abstractmethod

# Iterator Interface
class Iterator(ABC):
    @abstractmethod
    def has_next(self):
        pass
    @abstractmethod
    def next(self):
        pass


# Aggregate Interface
class Library(ABC):
    @abstractmethod
    def create_iterator(self):
        pass


# Concrete Iterator
class BookIterator(Iterator):
    def __init__(self, books):
        self._books = books
        self._index = 0
    def has_next(self):
        return self._index < len(self._books)
    def next(self):
        if self.has_next():
            book = self._books[self._index]
            self._index += 1
            return book
        return None


# Concrete Aggregate
class BookLibrary(Library):
    def __init__(self):
        self._books = []
    def add_book(self, book):
        self._books.append(book)
    def create_iterator(self):
        return BookIterator(self._books)


# Client Code
library = BookLibrary()
library.add_book("Python Programming")
library.add_book("Data Structures")
library.add_book("Operating Systems")
library.add_book("Machine Learning")

iterator = library.create_iterator()

while iterator.has_next():
    print("Book:", iterator.next())
```

#Q4

A company builds an online team chat application (similar to Slack /
WhatsApp groups / MS Teams).

Features to be supported:
- Users join chat rooms
- Send messages to group
- Private messaging
- Notifications
- User join/leave alerts

```python
# MEDIATOR PATTERN

from abc import ABC, abstractmethod

class ChatMediator(ABC):
    @abstractmethod
    def send_message(self, message, user):
        pass
    @abstractmethod
    def send_private_message(self, message, sender, receiver):
        pass


class ChatRoom(ChatMediator):
    def __init__(self):
        self.users = []

    # User joins the chat room
    def add_user(self, user):
        self.users.append(user)
        print(user.name, "joined the chat room")
        # Notify other users
        for u in self.users:
            if u != user:
                u.receive_notification(user.name + " joined the chat room")

    # User leaves the chat room
    def remove_user(self, user):
        if user in self.users:
            self.users.remove(user)
            print(user.name, "left the chat room")
            # Notify other users
            for u in self.users:
                u.receive_notification(user.name + " left the chat room")

    # Group message
    def send_message(self, message, sender):
        for user in self.users:
            if user != sender:
                user.receive(message, sender)

    # Private message
    def send_private_message(self, message, sender, receiver):
        if receiver in self.users:
            receiver.receive_private(message, sender)



# Colleague Class
class User:
    def __init__(self, name, mediator):
        self.name = name
        self.mediator = mediator

    def send(self, message):
        print(f"{self.name} sends: {message}")
        self.mediator.send_message(message, self)

    def send_private(self, message, receiver):
        print(f"{self.name} sends private message to {receiver.name}: {message}")
        self.mediator.send_private_message(message, self, receiver)

    def receive(self, message, sender):
        print(f"{self.name} received from {sender.name}: {message}")

    def receive_private(self, message, sender):
        print(f"{self.name} received private message from {sender.name}: {message}")

    def receive_notification(self, message):
        print(f"Notification for {self.name}: {message}")


# Client Code
chat_room = ChatRoom()
user1 = User("Alice", chat_room)
user2 = User("Bob", chat_room)
user3 = User("Charlie", chat_room)

# Users join
chat_room.add_user(user1)
print("")
chat_room.add_user(user2)
print("")
chat_room.add_user(user3)

# Group message
print("\n--- Group Message ---")
user1.send("Hello everyone!")

# Private message
print("\n--- Private Message ---")
user2.send_private("Hi Alice!", user1)

# User leaves
print("\n--- User Leaving ---")
chat_room.remove_user(user3)
```

#Q5

Consider a video game (Adventure / Racing / Shooting).

While playing, the player progresses through:
- Levels
- Missions
- Health status
- Weapons collected
- Score points
- Map position

Players expect features like:
- Save Game
- Load Game
- Checkpoints
- Restart from last saved state

To implement this, the system must capture and restore the player’s state
without exposing internal game logic.

```python
# MEMENTO PATTERN

# Memento Class
class GameState:
    def __init__(self, level, health, score, position):
        self._level = level
        self._health = health
        self._score = score
        self._position = position
    def get_state(self):
        return (
            self._level,
            self._health,
            self._score,
            self._position
        )


# Originator
class Game:
    def __init__(self):
        self._level = 1
        self._health = 100
        self._score = 0
        self._position = "Start"

    def play(self, level, health, score, position):
        self._level = level
        self._health = health
        self._score = score
        self._position = position

    def save(self):
        return GameState(
            self._level,
            self._health,
            self._score,
            self._position
        )

    def restore(self, state):
        (
            self._level,
            self._health,
            self._score,
            self._position
        ) = state.get_state()

    def show_state(self):
        print(
            self._level,
            self._health,
            self._score,
            self._position
        )


# Caretaker
class GameHistory:
    def __init__(self):
        self._states = []

    def save_state(self, state):
        self._states.append(state)

    def load_state(self):
        if not self._states:
            return None
        return self._states.pop()


# Client Code
game = Game()
history = GameHistory()

game.play(2, 80, 1000, "Forest")
history.save_state(game.save())

game.play(3, 50, 2000, "Castle")

print("Current Game:")
game.show_state()

game.restore(history.load_state())

print("After Loading:")
game.show_state()
```

#Q6

Consider a Stock Market Monitoring Application used by investors.

Features:
- Investors subscribe to stock updates
- Receive alerts when price changes
- Track multiple stocks
- Get real-time notifications

Example:
- If TCS price rises → Investors get alert
- If Infosys falls → Subscribers get notification

```python
# OBSERVER PATTERN

from abc import ABC, abstractmethod

# Observer
class Investor(ABC):
    @abstractmethod
    def update(self, stock, price):
        pass

# Concrete Observer
class StockInvestor(Investor):
    def __init__(self, name):
        self.name = name
    def update(self, stock, price):
        print(f"{self.name} received alert: {stock} price is now Rs. {price}")

# Subject
class StockMarket:
    def __init__(self):
        self.investors = []
        self.stock_prices = {}

    def subscribe(self, investor):
        self.investors.append(investor)

    def unsubscribe(self, investor):
        if investor in self.investors:
            self.investors.remove(investor)

    def update_price(self, stock, price):
        self.stock_prices[stock] = price
        print(f"\n{stock} price updated to Rs. {price}")

        # Notify all investors
        for investor in self.investors:
            investor.update(stock, price)


# Client
market = StockMarket()

investor1 = StockInvestor("Alice")
investor2 = StockInvestor("Bob")

market.subscribe(investor1)
market.subscribe(investor2)

market.update_price("TCS", 3500)
market.update_price("Infosys", 1800)

market.unsubscribe(investor2)

market.update_price("TCS", 3600)
```

#Q7

Consider an E-commerce platform (like Amazon / Flipkart).
When a customer places an order, it goes through multiple states:
1. Order Placed
2. Payment Confirmed
3. Shipped
4. Out for Delivery
5. Delivered
6. Cancelled / Returned

The system’s behaviour changes depending on the current order state.

Example:
- You can cancel only before shipping
- You can track only after shipping
- You can return only after delivery

```python
# STATE PATTERN

from abc import ABC, abstractmethod

# State
class OrderState(ABC):
    @abstractmethod
    def cancel(self, order):
        pass
    @abstractmethod
    def track(self, order):
        pass
    @abstractmethod
    def return_order(self, order):
        pass


# Concrete States
class OrderPlaced(OrderState):
    def cancel(self, order):
        print("Order cancelled")
        order.state = Cancelled()
    def track(self, order):
        print("Cannot track. Order has not been shipped.")
    def return_order(self, order):
        print("Cannot return. Order has not been delivered.")


class PaymentConfirmed(OrderState):
    def cancel(self, order):
        print("Order cancelled")
        order.state = Cancelled()
    def track(self, order):
        print("Cannot track. Order has not been shipped.")
    def return_order(self, order):
        print("Cannot return. Order has not been delivered.")


class Shipped(OrderState):
    def cancel(self, order):
        print("Cannot cancel. Order has been shipped.")
    def track(self, order):
        print("Order is in transit")
    def return_order(self, order):
        print("Cannot return. Order has not been delivered.")


class OutForDelivery(OrderState):
    def cancel(self, order):
        print("Cannot cancel. Order is out for delivery.")
    def track(self, order):
        print("Order is out for delivery")
    def return_order(self, order):
        print("Cannot return. Order has not been delivered.")


class Delivered(OrderState):
    def cancel(self, order):
        print("Cannot cancel. Order has been delivered.")
    def track(self, order):
        print("Order was delivered")
    def return_order(self, order):
        print("Return request accepted")
        order.state = Returned()


class Cancelled(OrderState):
    def cancel(self, order):
        print("Order is already cancelled")
    def track(self, order):
        print("Cannot track. Order is cancelled.")
    def return_order(self, order):
        print("Cannot return. Order is cancelled.")


class Returned(OrderState):
    def cancel(self, order):
        print("Order has already been returned")
    def track(self, order):
        print("Order has been returned")
    def return_order(self, order):
        print("Order is already returned")


# Context
class Order:
    def __init__(self):
        self.state = OrderPlaced()
    def cancel(self):
        self.state.cancel(self)
    def track(self):
        self.state.track(self)
    def return_order(self):
        self.state.return_order(self)
    def set_state(self, state):
        self.state = state


# Client
order = Order()

print("Order Placed:")
order.track()
order.cancel()

print("\nNew Order:")
order = Order()

print("Payment Confirmed:")
order.set_state(PaymentConfirmed())

print("Shipped:")
order.set_state(Shipped())
order.track()
order.cancel()

print("\nOut for Delivery:")
order.set_state(OutForDelivery())
order.track()

print("\nDelivered:")
order.set_state(Delivered())
order.return_order()
```

#Q8

Consider an E-commerce application.

When a customer checks out, they can choose different payment methods:
- Credit Card
- Debit Card
- UPI
- Net Banking
- Wallets

Each payment method has different processing logic, but the checkout flow
remains the same.

```python
# STRATEGY PATTERN

from abc import ABC, abstractmethod

# Strategy
class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


# Concrete Strategies
class CreditCardPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Credit Card")

class DebitCardPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Debit Card")

class UPIPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using UPI")

class NetBankingPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Net Banking")

class WalletPayment(PaymentStrategy):
    def pay(self, amount):
        print(f"Paid Rs. {amount} using Wallet")


# Context
class Checkout:
    def __init__(self, payment_strategy):
        self.payment_strategy = payment_strategy
    def set_payment_strategy(self, payment_strategy):
        self.payment_strategy = payment_strategy
    def checkout(self, amount):
        print("Processing order...")
        self.payment_strategy.pay(amount)
        print("Order placed successfully")


# Client
payment_method = Checkout(CreditCardPayment())
payment_method.checkout(2500)
print("")
payment_method.set_payment_strategy(WalletPayment())
payment_method.checkout(800)
```

Consider an Online Food Delivery System (Swiggy / Zomato).

When a customer places an order, the system follows a fixed workflow,
regardless of the restaurant:
1. Accept order
2. Assign delivery partner
3. Deliver order
4. Collect feedback

Even though the flow is fixed, the following steps differs:
- Preparation time and method (Based on the type of restaurant)
- Packing food (Cloud kitchens use automated packing)

```python
# TEMPLATE PATTERN

from abc import ABC, abstractmethod

# Abstract Class
class FoodDelivery(ABC):
    # Template method
    def place_order(self):
        self.accept_order()
        self.prepare_food()
        self.pack_food()
        self.assign_delivery_partner()
        self.deliver_order()
        self.collect_feedback()

    def accept_order(self):
        print("Order accepted")

    @abstractmethod
    def prepare_food(self):
        pass

    @abstractmethod
    def pack_food(self):
        pass

    def assign_delivery_partner(self):
        print("Delivery partner assigned")

    def deliver_order(self):
        print("Order delivered")

    def collect_feedback(self):
        print("Feedback collected")


# Concrete Class
class RegularRestaurant(FoodDelivery):
    def prepare_food(self):
        print("Restaurant prepares food manually")
    def pack_food(self):
        print("Restaurant packs food manually")

class CloudKitchen(FoodDelivery):
    def prepare_food(self):
        print("Cloud kitchen prepares food using automated process")
    def pack_food(self):
        print("Food packed using automated packing")


# Client
print("Regular Restaurant:")
order1 = RegularRestaurant()
order1.place_order()

print("\nCloud Kitchen:")
order2 = CloudKitchen()
order2.place_order()
```

#Q10

Consider a Hospital Management System where different types of
patients are admitted:
- General Patients
- Children Patients
- Senior Citizens

Hospital management needs to perform different operations on patients, such
as:
- Doctor consultation
- Health checkup
- Insurance claim processing
- Medicine prescription
- Billing calculation

Instead of adding all these operations inside patient classes, the application
should use a specific design pattern to separate operations from objects

```python
# VISITOR PATTERN

from abc import ABC, abstractmethod

# Visitor
class HospitalVisitor(ABC):
    @abstractmethod
    def visit_general(self, patient):
        pass
    @abstractmethod
    def visit_child(self, patient):
        pass
    @abstractmethod
    def visit_senior(self, patient):
        pass


# Concrete Elements
class GeneralPatient:
    def accept(self, visitor):
        visitor.visit_general(self)

class ChildPatient:
    def accept(self, visitor):
        visitor.visit_child(self)

class SeniorPatient:
    def accept(self, visitor):
        visitor.visit_senior(self)


# Concrete Visitor
class BillingVisitor(HospitalVisitor):
    def visit_general(self, patient):
        print("General patient billing: Rs. 1000")
    def visit_child(self, patient):
        print("Child patient billing: Rs. 800")
    def visit_senior(self, patient):
        print("Senior citizen billing: Rs. 700")


# Client
general = GeneralPatient()
child = ChildPatient()
senior = SeniorPatient()
billing = BillingVisitor()

general.accept(billing)
child.accept(billing)
senior.accept(billing)
```

#Q11

Consider a Bank Loan Processing System.

When a customer applies for a loan, the request passes through multiple
approval levels:
1. Clerk → Verifies documents
2. Manager → Approves medium loans
3. Regional Manager → Approves large loans
4. Head Office → Approves very large loans

Each authority has a loan approval limit.
If one level cannot approve, the request is forwarded to the next level.

```python
# CHAIN OF RESPONSIBILITY PATTERN

from abc import ABC, abstractmethod

# Handler
class LoanApprover(ABC):
    def __init__(self, next_approver=None):
        self.next_approver = next_approver
    def set_next_approver(self, next_approver):
        self.next_approver = next_approver
    @abstractmethod
    def approve(self, amount):
        pass


# Concrete Handlers
class Clerk(LoanApprover):
    def approve(self, amount):
        if amount <= 100000:
            print(f"Clerk approved loan of Rs. {amount}")
        elif self.next_approver:
            self.next_approver.approve(amount)

class Manager(LoanApprover):
    def approve(self, amount):
        if amount <= 500000:
            print(f"Manager approved loan of Rs. {amount}")
        elif self.next_approver:
            self.next_approver.approve(amount)

class RegionalManager(LoanApprover):
    def approve(self, amount):
        if amount <= 2000000:
            print(f"Regional Manager approved loan of Rs. {amount}")
        elif self.next_approver:
            self.next_approver.approve(amount)

class HeadOffice(LoanApprover):
    def approve(self, amount):
        print(f"Head Office approved loan of Rs. {amount}")


# Client
clerk = Clerk()
manager = Manager()
regional = RegionalManager()
head = HeadOffice()

clerk.set_next_approver(manager)
manager.set_next_approver(regional)
regional.set_next_approver(head)

clerk.approve(50000)
clerk.approve(7000000)
```
