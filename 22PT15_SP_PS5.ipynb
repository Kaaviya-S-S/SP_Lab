````python
# Long Method → Extract Method

class Order:
  def calculate_bill(self, price, quantity, customer_type):
    total = self.calculate_subtotal(price, quantity)
    total = self.apply_discount(total, customer_type)
    total = self.add_tax(total)
    total = self.add_shipping(total)
    return total

  def calculate_subtotal(self, price, quantity):
    return price * quantity

  def apply_discount(self, total, customer_type):
    if customer_type == "regular":
        discount = total * 0.05
    elif customer_type == "premium":
        discount = total * 0.10
    else:
        discount = 0
    return total - discount

  def add_tax(self, total):
    tax = total * 0.18
    return total + tax

  def add_shipping(self, total):
    if total > 5000:
        shipping = 0
    else:
        shipping = 100
    return total + shipping


order = Order()
print("Final bill: ", order.calculate_bill(100, 2, "premium"))
````

#Q2

Moving features between objects
```
class Student:
  init_(self, name, marks, department):
    self.name = name
    self.marks = marks
    self.department = department
  def department_average(self, students):
    total = 0
    count = 0
    for s in students:
      if s.department == self.department:
        total += s.marks
        count += 1
    return total / count

```

````python
# Feature Envy → Move Method

class Student:
  def __init__(self, name, marks, department):
    self.name = name
    self.marks = marks
    self.department = department

class Department:
  def __init__(self, name, students):
    self.name = name
    self.students = students

  def department_average(self):
    total = 0
    count = 0
    for s in self.students:
        total += s.marks
        count += 1
    return total / count if count > 0 else 0


s1 = Student("Arun", 80, "CSE")
s2 = Student("Priya", 90, "CSE")
s3 = Student("Ravi", 70, "ECE")

cse = Department("CSE", [s1, s2])
print("Department Average: ", cse.department_average())
````

```
class BonusCalculator:
  def calculate_bonus(self, salary, years, rating):
    if salary > 50000:
      if years > 5:
        if rating == "excellent":
          return salary * 0.2
        else:
          return salary * 0.1
      else:
        if rating == "excellent":
          return salary * 0.15
        else:

    else:

```

````python
# Complex Conditional / Nested Conditionals → Guard Clauses / Simplify Conditional Expression

class BonusCalculator:
  def calculate_bonus(self, salary, years, rating):
    if salary <= 50000:
      return salary * 0.02

    if years > 5:
      if rating == "excellent":
        return salary * 0.2
      return salary * 0.1

    if rating == "excellent":
      return salary * 0.15

    return salary * 0.05

bonus_calculator = BonusCalculator()
bonus_calculator.calculate_bonus(80000, 8, "excellent")
````
