#Q1

In many applications (e.g. web servers, APIs, config loaders), you want to
store frequently used data in memory and access it globally — but you
don’t want multiple cache objects floating around. Ensure only one
cache instance exists throughout the app.

```python
# -- SINGLETON --

class MemoryCache:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
          cls._instance = super(MemoryCache, cls).__new__(cls)
          cls._instance.storage = {}
        return cls._instance

    def set_data(self, key, value):
        self.storage[key] = value
        print(f"Cached: {key} = {value}")

    def get_data(self, key):
        return self.storage.get(key, "Data not found")

    def clear_cache(self):
        self.storage.clear()
        print("Cache cleared.")

# Application Component A (e.g., Auth Service)
cache_a = MemoryCache()
cache_a.set_data("api_token", "12345-ABCDE")

# Application Component B (e.g., API Handler)
cache_b = MemoryCache()
print(f"Component B retrieved token: {cache_b.get_data('api_token')}")

print(f"\nInstance A ID: {id(cache_a)}")
print(f"Instance B ID: {id(cache_b)}")

if cache_a is cache_b:
    print("TRUE: Both components are using the exact same Cache instance.")
```

#Q2

A restaurant offers:
* Vegetarian meal set
* Non-vegetarian meal set

Each set includes:
* Starter
* Main course
* Dessert

```python
from abc import ABC, abstractmethod

# --- abstract product ---
class Starter(ABC):
    @abstractmethod
    def get_name(self):
        pass

class MainCourse(ABC):
    @abstractmethod
    def get_name(self):
        pass

class Dessert(ABC):
    @abstractmethod
    def get_name(self):
        pass


# --- concrete product (Vegetarian) ---
class VegetarianStarter(Starter):
    def get_name(self):
        return "abcd - Veg Starter"

class VegetarianMainCourse(MainCourse):
    def get_name(self):
        return "efgh - Veg Main Course"

class VegetarianDessert(Dessert):
    def get_name(self):
        return "ijkl - Veg Dessert"


# --- concrete product (Non-Vegetarian) ---
class NonVegetarianStarter(Starter):
    def get_name(self):
        return "mnop - Non-Veg Starter"

class NonVegetarianMainCourse(MainCourse):
    def get_name(self):
        return "qrst - Non-Veg Main Course"

class NonVegetarianDessert(Dessert):
    def get_name(self):
        return "uvwx - Non-Veg Dessert"


# --- abstract factory ---
class MealFactory(ABC):
    @abstractmethod
    def create_starter(self):
        pass

    @abstractmethod
    def create_main_course(self):
        pass

    @abstractmethod
    def create_dessert(self):
        pass


# --- concrete factory ---
class VegetarianMealFactory(MealFactory):
    def create_starter(self):
        return VegetarianStarter()

    def create_main_course(self):
        return VegetarianMainCourse()

    def create_dessert(self):
        return VegetarianDessert()

class NonVegetarianMealFactory(MealFactory):
    def create_starter(self):
        return NonVegetarianStarter()

    def create_main_course(self):
        return NonVegetarianMainCourse()

    def create_dessert(self):
        return NonVegetarianDessert()


def show_restaurant_menu(factory: MealFactory):
    starter = factory.create_starter()
    main = factory.create_main_course()
    dessert = factory.create_dessert()

    print(f"--- Serving ---")
    print(f"Starter:     {starter.get_name()}")
    print(f"Main Course: {main.get_name()}")
    print(f"Dessert:     {dessert.get_name()}\n")


non_veg_factory = NonVegetarianMealFactory()
show_restaurant_menu(non_veg_factory)

veg_factory = VegetarianMealFactory()
show_restaurant_menu(veg_factory)
```

# Q3

 A payment gateway supports different types of payments based on
customer choice
* Credit Card
* Debit Card
* UPI
* Net banking
* Wallets

```python
# -- FACTORY METHOD --

from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def process_payment(self):
        pass

class CreditCardPaymentGateway(PaymentGateway):
    def process_payment(self):
        print("Processing credit card payment...")

class DebitCardPaymentGateway(PaymentGateway):
    def process_payment(self):
        print("Processing debit card payment...")

class UPIPaymentGateway(PaymentGateway):
    def process_payment(self):
        print("Processing UPI payment...")

class NetBankingPaymentGateway(PaymentGateway):
    def process_payment(self):
        print("Processing net banking payment...")


class PaymentGatewayFactory:
  @staticmethod
  def get_payment_gateway(payment_type):
    if payment_type == "credit_card":
      return CreditCardPaymentGateway()
    elif payment_type == "debit_card":
      return DebitCardPaymentGateway()
    elif payment_type == "upi":
      return UPIPaymentGateway()
    elif payment_type == "net_banking":
      return NetBankingPaymentGateway()
    else:
      raise ValueError("Invalid payment type")


factory = PaymentGatewayFactory()

credit_card_gateway = factory.get_payment_gateway("credit_card")
credit_card_gateway.process_payment()

UPI_gateway = factory.get_payment_gateway("upi")
UPI_gateway.process_payment()
```

#Q4

An application needs to generate multiple covering letters to be sent to different recipients. The content of the covering letter is standards whereas the recipients are based on a dynamic list supplied by the user.

```python
# -- PROTOTYPE --

from abc import ABC, abstractmethod
import copy

class Document(ABC):
  @abstractmethod
  def clone(self):
    pass

class CoverLetter(Document):
  def __init__(self, content, recipient):
    self.content = content
    self.recipient = recipient

  def clone(self):
    return copy.deepcopy(self)

  def display(self):
    print("\nCovering Letter:")
    print(f"Content: {self.content}")
    print(f"Recipient: {self.recipient}")


original_covering_letter = CoverLetter(
    "The content of covering letter is....",
    "abc"
)

cloned_letter_1 = original_covering_letter.clone()
cloned_letter_1.recipient = "efgh"

cloned_letter_2 = original_covering_letter.clone()
cloned_letter_2.recipient = "ijkl"

original_covering_letter.display()
cloned_letter_1.display()
cloned_letter_2.display()
```

#Q5

CV Builder application – with sections to capture the following details
and build a complete CV :
* Education
* Experience
* Personal Details
* Specialisations
* Hobbies etc

```python
# -- BUILDER --

from abc import ABC, abstractmethod

# Product
class CV:
  def __init__(self):
    self.education = None
    self.experience = None
    self.personal_details = None
    self.specialisations = None
    self.hobbies = None

  def display(self):
    print("CV: ")
    print(f"Education: {self.education}")
    print(f"Experience: {self.experience}")
    print(f"Personal Details: {self.personal_details}")
    print(f"Specialisations: {self.specialisations}")
    print(f"Hobbies: {self.hobbies}")


# Builder
class CVBuilder(ABC):
  @abstractmethod
  def build_education(self):
    pass

  @abstractmethod
  def build_experience(self):
    pass

  @abstractmethod
  def build_personal_details(self):
    pass

  @abstractmethod
  def build_specialisations(self):
    pass

  @abstractmethod
  def build_hobbies(self):
    pass

  @abstractmethod
  def get_cv(self):
    pass


# Concrete Builder
class StudentCVBuilder(CVBuilder):
  def __init__(self):
    self.cv = CV()

  def build_education(self):
    self.cv.education = "M.Sc TCS"

  def build_experience(self):
    self.cv.experience = "abcd"

  def build_personal_details(self):
    self.cv.personal_details = "xxx"

  def build_specialisations(self):
    self.cv.specialisations = "yyy"

  def build_hobbies(self):
    self.cv.hobbies = "zzz"

  def get_cv(self):
    return self.cv


# Director
class CVBuildDirector:
  def __init__(self, cv_builder):
    self.cv_builder = cv_builder

  def construct_cv(self):
    self.cv_builder.build_education()
    self.cv_builder.build_experience()
    self.cv_builder.build_personal_details()
    self.cv_builder.build_specialisations()
    self.cv_builder.build_hobbies()


cv_builder = StudentCVBuilder()
cv_director = CVBuildDirector(cv_builder)
cv_director.construct_cv()
cv = cv_builder.get_cv()
cv.display()
```
