#Q1- Technical Debt


1. Existing code
```
class Cart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price, quantity):
        self.items.append((name, price, quantity))

    def total(self):
        total = 0
        for i in self.items:
            if i[0] == "book":
                total += i[1] * i[2] * 0.9
            elif i[0] == "pen":
                total += i[1] * i[2] * 0.95
            else:
                total += i[1] * i[2]
        return total
```

2. Design Smell
- Technical Debt – Hard-coded Business Logic
- The discount percentages for books (10%) and pens (5%) are directly hard-coded inside the total() method.

3. Problem with the design
- Business logic is mixed with the cart calculation.
- Adding a new product category requires modifying the existing total() method.
- Hard-coded discount values make the code difficult to maintain.
- Repeated changes can introduce bugs and increase maintenance costs.

4. Refactoring techniques
- Extract Method: Separate discount calculation from the total() method.
- Replace Magic Number with Symbolic Constant: Replace 0.9 and 0.95 with named constants.
- Move Method: Move discount-related logic into a separate class responsible for product discounts.

````python
# Discount class
class Discount:
    BOOK_DISCOUNT = 0.10
    PEN_DISCOUNT = 0.05

    def get_discount(self, name):
        if name == "book":
            return self.BOOK_DISCOUNT
        elif name == "pen":
            return self.PEN_DISCOUNT
        else:
            return 0


# Cart class
class Cart:
    def __init__(self):
        self.items = []
        self.discount = Discount()

    def add_item(self, name, price, quantity):
        self.items.append((name, price, quantity))

    def total(self):
        total = 0
        for name, price, quantity in self.items:
            discount = self.discount.get_discount(name)
            amount = price * quantity
            total += amount * (1 - discount)
        return total


# Main program
cart = Cart()

cart.add_item("book", 100, 2)
cart.add_item("pen", 20, 5)
cart.add_item("pencil", 10, 3)

print("Total Amount:", cart.total())
````

2. Design Smell

  - Adding books.
  - Issuing and returning books.
  - Calculating fines.

3. Problem with the design

4. Refactoring techniques

  - Move Method: Move methods to the appropriate classes.

````python
# Handles book operations
class Library:
    def add_book(self, title):
        print("Book added:", title)

    def issue_book(self, title, user):
        print("Book issued:", title, "to", user)

    def return_book(self, title):
        print("Book returned:", title)


# Handles fine calculation
class FineCalculator:
    def calculate_fine(self, days):
        return days * 5


# Handles email notifications
class EmailService:
    def send_email(self, user):
        print("Reminder email sent to:", user)


# Handles report generation
class ReportGenerator:
    def generate_report(self):
        print("Library report generated")


# Main program
library = Library()
fine = FineCalculator()
email = EmailService()
report = ReportGenerator()

library.add_book("Python Programming")
library.issue_book("Python Programming", "John")
library.return_book("Python Programming")

print("Fine:", fine.calculate_fine(3))

email.send_email("John")
report.generate_report()
````

#Q3 - Abstraction smells

1. Existing code
```
class Payment:

    def pay_by_card(self):
        print("Card payment")

    def pay_by_upi(self):
        print("UPI payment")

    def pay_by_cash(self):
        print("Cash payment")
```

2. Design Smell
- Missing Abstraction (Insufficient Abstraction)
- The Payment class directly implements different payment methods without providing a common abstraction for payment processing.

3. Problem with the design
- All payment methods are grouped into a single class.
- There is no common interface for different payment methods.
- Adding a new payment method requires modifying the existing class.
- Code becomes difficult to extend and maintain.

4. Refactoring techniques
- Extract Superclass: Create a common abstract class Payment.
- Replace Conditional with Polymorphism: Allow each payment method to implement its own payment operation.
- Move Method: Move individual payment operations into their respective classes.

````python
from abc import ABC, abstractmethod

# Abstract class
class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass

# Card payment
class CardPayment(Payment):
    def pay(self):
        print("Card payment")

# UPI payment
class UPIPayment(Payment):
    def pay(self):
        print("UPI payment")

# Cash payment
class CashPayment(Payment):
    def pay(self):
        print("Cash payment")

# Main program
def process_payment(payment):
    payment.pay()

process_payment(CardPayment())
process_payment(UPIPayment())
process_payment(CashPayment())
````

#Q4 - Encapsulation smells

1. Existing code
```
class Student:

    def __init__(self):
        self.name = ""
        self.marks = 0
        self.grade = ""


s = Student()
s.name = "Ali"
s.marks = -50
s.grade = "A"
```

2. Design Smell
- Data Class / Poor Encapsulation
- The class exposes its attributes (name, marks, grade) directly, allowing external code to modify them without validation.

3. Problem with the design
- Attributes are publicly accessible.
- Invalid values can be assigned, such as negative marks.
- There is no validation to ensure data consistency.
- The internal state of the object can be modified directly.
- It becomes difficult to maintain data integrity.

4. Refactoring techniques
- Encapsulate Field: Make attributes private.
- Self Encapsulate Field: Access attributes through getter and setter methods.
-  Add Validation: Validate values before modifying the attributes.

````python
class Student:
    def __init__(self):
        self.name = ""
        self.marks = 0
        self.grade = ""

    # Getter and setter for name
    def get_name(self):
        return self.name
    def set_name(self, name):
        self.name = name

    # Getter and setter for marks
    def get_marks(self):
        return self.marks
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.marks = marks
        else:
            raise ValueError("Marks must be between 0 and 100")

    # Getter and setter for grade
    def get_grade(self):
        return self.grade
    def set_grade(self, grade):
        self.grade = grade


s = Student()
s.set_name("Ali")
s.set_marks(85)
s.set_grade("A")

print("Name:", s.get_name())
print("Marks:", s.get_marks())
print("Grade:", s.get_grade())
````

#Q5 - Modularization smells

1. Existing code
```
class Application:
    def login(self, username, password):
        if username == "admin" and password == "123":
            print("Login success")
        else:
            print("Login failed")

    def save_user(self, username):
        print("Saving user to database")

    def display_dashboard(self):
        print("Welcome to dashboard")
```

2. Design Smell
- Low Cohesion / Poor Modularization
- The Application class handles multiple unrelated responsibilities, such as authentication, database operations and dashboard display.

3. Problem with the design
- Different functionalities are grouped into a single class.
- Authentication, database and user interface logic are mixed together.
- Changes in one functionality may affect other functionalities.
- Code becomes difficult to maintain, test and reuse.
- The class has low cohesion because its methods perform unrelated tasks.

4. Refactoring techniques
- Extract Class: Divide the Application class into separate classes based on their responsibilities.
- Move Method: Move each method into its appropriate class.
- Extract Module: Organize related classes into separate modules.

````python
# Authentication module
class Authentication:
    def login(self, username, password):
        if username == "admin" and password == "123":
            print("Login success")
            return True
        else:
            print("Login failed")
            return False

# Database module
class UserDatabase:
    def save_user(self, username):
        print("Saving user to database:", username)

# Dashboard module
class Dashboard:
    def display_dashboard(self):
        print("Welcome to dashboard")

# Application class
class Application:
    def __init__(self):
        self.auth = Authentication()
        self.database = UserDatabase()
        self.dashboard = Dashboard()
    def run(self, username, password):
        if self.auth.login(username, password):
            self.database.save_user(username)
            self.dashboard.display_dashboard()

app = Application()
app.run("admin", "123")
````

#Q6 - Hierarchy Smells


1. Existing code
```
class Bird:
    def fly(self):
        print("Flying")

class Sparrow(Bird):
    pass
class Penguin(Bird):
    pass

p = Penguin()
p.fly()
```

2. Design Smell
- Incorrect Inheritance
- The Penguin class inherits the fly() method from the Bird class, even though penguins cannot fly. This means the child class inherits behaviour that it does not need or cannot support.

3. Problem with the design
- The inheritance hierarchy is incorrectly designed.
- Penguin inherits the fly() method unnecessarily.
- Calling p.fly() produces incorrect behaviour.
- It violates the Liskov Substitution Principle (LSP).
- Adding more bird types may make the inheritance hierarchy difficult to maintain.

4. Refactoring techniques
- Replace Inheritance with Delegation: Avoid forcing all bird types to inherit flying behaviour.
- Extract Superclass: Create a common Bird superclass containing only common bird properties and behaviours.
- Extract Interface: Introduce a separate Flyable interface for birds that can fly.

````python
from abc import ABC, abstractmethod

# Common abstract class
class Bird(ABC):
    @abstractmethod
    def move(self):
        pass

# Flying behaviour interface
class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass

# Sparrow can fly
class Sparrow(Bird, Flyable):
    def move(self):
        print("Sparrow is moving")
    def fly(self):
        print("Sparrow is flying")

# Penguin cannot fly
class Penguin(Bird):
    def move(self):
        print("Penguin is walking")


s = Sparrow()
s.fly()
s.move()

p = Penguin()
p.move()
````
